"""
BananaVision — Module A : Architecture CNN (Transfer Learning)
PHP Plantations du Haut-Penja — UCAC-ICAM 2026

Ce module implémente les deux architectures comparées :
  1. MobileNetV2  (Google, 2018) — léger, rapide, temps réel
  2. EfficientNet-B0 (Google Brain, 2019) — plus précis, compound scaling

STRATÉGIE DE TRANSFER LEARNING CHOISIE :
  Phase 1 — Feature Extraction (Epochs 1-10) :
    → Toutes les couches pré-entraînées GELÉES (requires_grad=False)
    → Seule la tête de classification est entraînée
    → Learning rate élevé (1e-3) : la tête part de zéro
    POURQUOI : éviter de "désapprendre" les features ImageNet dès le début

  Phase 2 — Fine-tuning (Epochs 11-20) :
    → Les derniers blocs du backbone sont DÉGELÉS
    → Learning rate très faible (1e-5) : ajustements fins seulement
    POURQUOI : adapter les features de haut niveau aux spécificités
    des bananes du Mungo (éclairage pack-house, fond tapis convoyeur)

JUSTIFICATION DU CHOIX ARCHITECTUREL POUR LE RAPPORT :
  - VGG-16 : 138M paramètres → overfitting certain sur < 10 000 images
  - ResNet-50 : 25M params → acceptable mais temps d'inférence > 100ms
  - MobileNetV2 : 3.4M params, 30ms → compatible convoyeur PHP (5s/banane)
  - EfficientNet-B0 : 5.3M params, 50ms → meilleur accuracy/taille ratio
"""

```python


import torch
import torch.nn as nn
from torchvision import models
from typing import Literal


print(f"CUDA disponible : {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU : {torch.cuda.get_device_name(0)}")


# ─────────────────────────────────────────────────────────────────
# BUILDER PRINCIPAL
# ─────────────────────────────────────────────────────────────────

def build_model(
    architecture: Literal["mobilenet_v2", "efficientnet_b0"],
    num_classes: int,
    task: Literal["binary", "multiclass"] = "binary",
    freeze_backbone: bool = True,
) -> nn.Module:
    """
    Construit et retourne le modèle CNN avec tête de classification adaptée.

    Args:
        architecture: "mobilenet_v2" ou "efficientnet_b0"
        num_classes: 2 pour binaire, N pour multi-classes
        task: "binary" (sigmoid) ou "multiclass" (softmax implicite dans CrossEntropy)
        freeze_backbone: True = Phase 1 (feature extraction)
                         False = Phase 2 (fine-tuning)
    Returns:
        nn.Module : modèle PyTorch prêt pour l'entraînement
    """
    if architecture == "mobilenet_v2":
        return _build_mobilenet_v2(num_classes, task, freeze_backbone)
    elif architecture == "efficientnet_b0":
        return _build_efficientnet_b0(num_classes, task, freeze_backbone)
    else:
        raise ValueError(f"Architecture inconnue : {architecture}. Choisir 'mobilenet_v2' ou 'efficientnet_b0'")


def _build_mobilenet_v2(num_classes: int, task: str, freeze_backbone: bool) -> nn.Module:
    """
    MobileNetV2 avec tête de classification PHP personnalisée.

    ARCHITECTURE INTERNE MobileNetV2 :
    Input (3×224×224)
    → Conv2d initiale
    → 17 Blocs Inversés Résiduels (InvertedResidual)
      [c'est ici que résident les features visuelles génériques d'ImageNet]
    → Conv2d finale → AdaptiveAvgPool → Dropout
    → Linear(1280 → num_classes)  ← CETTE COUCHE SEULEMENT est remplacée

    BLOC INVERSÉ RÉSIDUEL (différence clé avec ResNet) :
    Au lieu d'élargir puis rétrécir les canaux (ResNet),
    MobileNetV2 rétrécit d'abord (bottleneck), applique une conv depthwise
    (1 filtre par canal, très efficace), puis ré-élargit.
    Résultat : ~10x moins d'opérations que ResNet pour une qualité similaire.
    """
    # Chargement des poids pré-entraînés sur ImageNet
    weights = models.MobileNet_V2_Weights.IMAGENET1K_V1
    model = models.mobilenet_v2(weights=weights)

    # Gel du backbone
    if freeze_backbone:
        for param in model.parameters():
            param.requires_grad = False
        print("[MobileNetV2] Backbone gelé — seule la tête sera entraînée")
    else:
        # Phase 2 : dégeler les 5 derniers blocs (features de haut niveau)
        # Les premiers blocs restent gelés : ils détectent des primitives
        # (bords, coins) qui sont universelles et n'ont pas besoin d'adaptation
        _unfreeze_last_n_layers(model.features, n=5)
        print("[MobileNetV2] Fine-tuning : 5 derniers blocs dégelés")

    # Remplacement de la tête de classification
    # model.classifier[1] est le Linear(1280 → 1000) d'ImageNet
    in_features = model.classifier[1].in_features  # = 1280

    model.classifier = _build_head(in_features, num_classes, task)

    _print_trainable_params(model, "MobileNetV2")
    return model


def _build_efficientnet_b0(num_classes: int, task: str, freeze_backbone: bool) -> nn.Module:
    """
    EfficientNet-B0 avec tête de classification PHP personnalisée.

    COMPOUND SCALING — ce qui rend EfficientNet unique :
    Les architectures classiques agrandissent le réseau selon UNE dimension :
    - Profondeur (plus de couches) → ResNet, VGG
    - Largeur (plus de filtres par couche) → Wide ResNet
    - Résolution (images plus grandes) → pas standard

    EfficientNet prouve mathématiquement qu'il faut scaler les TROIS ensemble
    avec un ratio constant (φ = coefficient de scaling) :
    - depth    d = α^φ  (α = 1.2 pour B0)
    - width    w = β^φ  (β = 1.1)
    - resolution r = γ^φ (γ = 1.15)

    EfficientNet-B0 est le modèle de base (φ=1). B1 à B7 sont des versions
    plus grandes. B0 est optimal pour notre petit dataset PHP.
    """
    weights = models.EfficientNet_B0_Weights.IMAGENET1K_V1
    model = models.efficientnet_b0(weights=weights)

    if freeze_backbone:
        for param in model.parameters():
            param.requires_grad = False
        print("[EfficientNet-B0] Backbone gelé")
    else:
        _unfreeze_last_n_layers(model.features, n=3)
        print("[EfficientNet-B0] Fine-tuning : 3 derniers blocs dégelés")

    # EfficientNet-B0 : sortie du backbone = 1280 features (même que MobileNetV2)
    in_features = model.classifier[1].in_features  # = 1280

    model.classifier = _build_head(in_features, num_classes, task)

    _print_trainable_params(model, "EfficientNet-B0")
    return model


# ─────────────────────────────────────────────────────────────────
# FONCTIONS UTILITAIRES
# ─────────────────────────────────────────────────────────────────

def _build_head(in_features: int, num_classes: int, task: str) -> nn.Sequential:
    """
    Construit la tête de classification adaptée à la tâche PHP.

    POURQUOI Dropout(0.3) ?
    Avec ~5000 images, le modèle risque de mémoriser les exemples d'entraînement.
    Le Dropout désactive aléatoirement 30% des neurones à chaque passe
    → force le réseau à apprendre des représentations robustes et redondantes.

    BINAIRE vs MULTI-CLASSES :
    - Binaire : 1 neurone de sortie + sigmoid → probabilité [0,1] d'être "rejeté"
      Utilisé avec BCEWithLogitsLoss (numériquement stable)
    - Multi-classes : N neurones + pas d'activation → probas via softmax implicite
      Utilisé avec CrossEntropyLoss (= LogSoftmax + NLLLoss)
    """
    if task == "binary":
        return nn.Sequential(
            nn.Dropout(p=0.3),
            nn.Linear(in_features, 256),
            nn.ReLU(),
            nn.Dropout(p=0.2),
            nn.Linear(256, 1),   # 1 seul neurone pour binaire
            # PAS de sigmoid ici ! BCEWithLogitsLoss l'inclut, plus stable
        )
    else:  # multiclass
        return nn.Sequential(
            nn.Dropout(p=0.3),
            nn.Linear(in_features, 256),
            nn.ReLU(),
            nn.Dropout(p=0.2),
            nn.Linear(256, num_classes),  # N neurones pour N classes
            # PAS de softmax ici ! CrossEntropyLoss l'inclut
        )


def _unfreeze_last_n_layers(feature_extractor: nn.Sequential, n: int) -> None:
    """Dégèle les n derniers blocs du backbone pour le fine-tuning."""
    layers = list(feature_extractor.children())
    for layer in layers[-n:]:
        for param in layer.parameters():
            param.requires_grad = True


def _print_trainable_params(model: nn.Module, name: str) -> None:
    """Affiche le nombre de paramètres entraînables vs total."""
    total  = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"[{name}] Params total: {total:,} | Entraînables: {trainable:,} "
          f"({100*trainable/total:.1f}%)")


def get_optimizer(model: nn.Module, phase: int = 1) -> torch.optim.Optimizer:
    """
    Retourne l'optimiseur adapté à la phase d'entraînement.

    Phase 1 (backbone gelé) : Adam lr=1e-3 sur la tête seulement
    Phase 2 (fine-tuning)   : Adam avec lr différencié :
        - Backbone dégelé : lr=1e-5 (très faible → ajustements fins)
        - Tête            : lr=1e-4 (modéré → la tête continue à apprendre)

    POURQUOI Adam et pas SGD ?
    Adam adapte le learning rate par paramètre → converge plus vite sur
    les petits datasets. SGD avec momentum est supérieur sur très grands
    datasets (ImageNet complet) mais requiert plus d'epochs pour converger.
    """
    if phase == 1:
        params = filter(lambda p: p.requires_grad, model.parameters())
        return torch.optim.Adam(params, lr=1e-3, weight_decay=1e-4)
    else:
        # Learning rates différenciés backbone vs tête
        backbone_params = []
        head_params = []
        for name, param in model.named_parameters():
            if param.requires_grad:
                if "classifier" in name:
                    head_params.append(param)
                else:
                    backbone_params.append(param)
        return torch.optim.Adam([
            {"params": backbone_params, "lr": 1e-5},
            {"params": head_params,     "lr": 1e-4},
        ], weight_decay=1e-4)
```

"""
BananaVision — Module A : Prétraitement des données
PHP Plantations du Haut-Penja — UCAC-ICAM 2026

Ce module gère :
  - Le chargement du dataset depuis un répertoire structuré
  - Le redimensionnement et la normalisation des images
  - L'augmentation de données pour compenser le petit dataset
  - La création des DataLoaders pour l'entraînement

Structure attendue du dataset :
    dataset/
        export/          ← images bananes conformes (Catégorie Extra ou I)
        rejete/          ← images bananes non conformes (Hors catégorie)
        defaut_mur/      ← bananes trop mûres (stade 6-7) → farine
        defaut_fongique/ ← cercosporiose, moisissures
        defaut_mecanique/← bruises, déformations
        defaut_calibre/  ← longueur < 14cm ou grade < 27mm
"""

```python


import os
from pathlib import Path

import torch
from torch.utils.data import DataLoader, random_split, WeightedRandomSampler
from torchvision import datasets, transforms
import numpy as np


# ─────────────────────────────────────────────────────────────────
# Constantes MobileNetV2 / EfficientNet-B0
# Les deux architectures attendent des images 224×224 normalisées
# avec les moyennes et écart-types d'ImageNet (car pré-entraînées dessus)
# ─────────────────────────────────────────────────────────────────
IMG_SIZE = 224
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

# Proportions train / validation / test
TRAIN_RATIO = 0.70
VAL_RATIO   = 0.15
TEST_RATIO  = 0.15


def get_transforms(mode: str = "train") -> transforms.Compose:
    """
    Retourne les transformations adaptées au mode.

    POURQUOI ces augmentations ?
    - RandomHorizontalFlip : une banane peut être filmée de gauche ou droite
      sur le convoyeur PHP → le modèle ne doit pas être sensible à l'orientation
    - RandomRotation(15°) : légère inclinaison possible sur le tapis
    - ColorJitter : simule les variations d'éclairage du pack-house
      (fluorescent en journée vs artificiel la nuit)
    - RandomResizedCrop : simule les bananes à différentes distances caméra
    ON N'UTILISE PAS : RandomVerticalFlip (une banane à l'envers n'existe pas
    sur un convoyeur), ni des rotations > 30° (irréaliste)

    Args:
        mode: "train" → augmentations actives | "val" ou "test" → aucune augmentation
    Returns:
        transforms.Compose : pipeline de transformations PyTorch
    """
    normalize = transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)

    if mode == "train":
        return transforms.Compose([
            # Redimensionnement légèrement plus grand puis crop aléatoire
            # → simule zoom variable caméra convoyeur
            transforms.RandomResizedCrop(IMG_SIZE, scale=(0.8, 1.0)),
            transforms.RandomHorizontalFlip(p=0.5),
            transforms.RandomRotation(degrees=15),
            transforms.ColorJitter(
                brightness=0.3,   # variations lumière pack-house PHP
                contrast=0.2,
                saturation=0.2,
                hue=0.05          # faible : on ne veut pas changer le vert en rouge
            ),
            transforms.ToTensor(),
            normalize,
        ])
    else:
        # Val et test : aucune augmentation, juste redimensionner et normaliser
        # IMPORTANT : on utilise les mêmes stats ImageNet → cohérence avec pré-entraînement
        return transforms.Compose([
            transforms.Resize(256),          # légèrement plus grand
            transforms.CenterCrop(IMG_SIZE), # crop central → zone de fruit
            transforms.ToTensor(),
            normalize,
        ])



# NOUVELLE FONCTION 

def load_binary_dataset(data_dir, batch_size=32, num_workers=2):
    """
    Charge les données depuis trois sous-dossiers : train, val, test.
    Aucun split aléatoire n'est effectué.
    
    Args:
        data_dir : chemin vers le dossier parent (ex: './dataset/binary')
                   Ce dossier doit contenir train/, val/, test/
        batch_size : taille des lots
        num_workers : nombre de processus pour le chargement
        balance : (non utilisé ici car on gère le déséquilibre via pos_weight)
    Returns:
        train_loader, val_loader, test_loader, class_names
    """
    from pathlib import Path
    from torch.utils.data import DataLoader
    from torchvision import datasets
    
    data_dir = Path(data_dir)
    
    # Chemins des trois sous-dossiers
    train_dir = data_dir / "train"
    val_dir   = data_dir / "val"
    test_dir  = data_dir / "test"
    
    # Vérifie que les dossiers existent
    for d in [train_dir, val_dir, test_dir]:
        if not d.exists():
            raise FileNotFoundError(f"Dossier introuvable : {d}")
    
    # Création des datasets avec les transformations adaptées
    train_set = datasets.ImageFolder(train_dir, transform=get_transforms("train"))
    val_set   = datasets.ImageFolder(val_dir,   transform=get_transforms("val"))
    test_set  = datasets.ImageFolder(test_dir,  transform=get_transforms("test"))
    
    class_names = train_set.classes
    print(f"[Binary] Classes détectées : {class_names}")
    print(f"[Binary] Train : {len(train_set)} images")
    print(f"[Binary] Val   : {len(val_set)} images")
    print(f"[Binary] Test  : {len(test_set)} images")
    
    # Création des DataLoaders
    # train : shuffle=True pour mélanger les batches
    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, num_workers=num_workers)
    val_loader   = DataLoader(val_set,   batch_size=batch_size, shuffle=False, num_workers=num_workers)
    test_loader  = DataLoader(test_set,  batch_size=batch_size, shuffle=False, num_workers=num_workers)
    
    # Note : l'argument `balance` n'est pas utilisé ici.
    # L'équilibrage est déjà fait via pos_weight dans la loss (BCEWithLogitsLoss)
    # Si tu veux un WeightedRandomSampler sur train, il faudrait l'ajouter, mais ce n'est pas indispensable.
    
    return train_loader, val_loader, test_loader, class_names


def load_multiclass_dataset(data_dir, batch_size=32, num_workers=2):
    """
    Charge les données multi-classes depuis trois sous-dossiers : train, val, test.
    """
    data_dir = Path(data_dir)
    train_dir = data_dir / "train"
    val_dir   = data_dir / "valid"
    test_dir  = data_dir / "test"

    for d in [train_dir, val_dir, test_dir]:
        if not d.exists():
            raise FileNotFoundError(f"Dossier introuvable : {d}")

    train_set = datasets.ImageFolder(train_dir, transform=get_transforms("train"))
    val_set   = datasets.ImageFolder(val_dir,   transform=get_transforms("val"))
    test_set  = datasets.ImageFolder(test_dir,  transform=get_transforms("test"))

    class_names = train_set.classes
    print(f"[Multiclass] Classes : {class_names}")
    print(f"[Multiclass] Train : {len(train_set)}, Val : {len(val_set)}, Test : {len(test_set)}")

    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True, num_workers=num_workers)
    val_loader   = DataLoader(val_set,   batch_size=batch_size, shuffle=False, num_workers=num_workers)
    test_loader  = DataLoader(test_set,  batch_size=batch_size, shuffle=False, num_workers=num_workers)

    return train_loader, val_loader, test_loader, class_names
```

BananaVision — Module A : Entraînement et Évaluation
PHP Plantations du Haut-Penja — UCAC-ICAM 2026

Ce module implémente :
  - La boucle d'entraînement (train_one_epoch)
  - L'évaluation complète avec toutes les métriques requises par le jury
  - La génération de la matrice de confusion commentée en termes métier PHP
  - La courbe ROC/AUC pour la tâche binaire
  - La mesure du temps d'inférence par image (ms)

MÉTRIQUES EXIGÉES PAR LE CAHIER DES CHARGES :
  ✓ Précision, Rappel, F1-score par classe
  ✓ Matrice de confusion commentée en termes métier PHP
  ✓ Courbe ROC et AUC (tâche binaire uniquement)
  ✓ Temps d'inférence moyen par image (ms)

```python


import time
from pathlib import Path

import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use("Agg")  # backend non-interactif pour serveur/Colab

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_recall_fscore_support,
    roc_curve,
    auc,
    ConfusionMatrixDisplay,
)


# ─────────────────────────────────────────────────────────────────
# BOUCLE D'ENTRAÎNEMENT
# ─────────────────────────────────────────────────────────────────

def train_one_epoch(
    model: nn.Module,
    loader: torch.utils.data.DataLoader,
    optimizer: torch.optim.Optimizer,
    criterion: nn.Module,
    device: torch.device,
    task: str = "binary",
) -> dict:
    """
    Exécute une epoch d'entraînement et retourne les métriques.

    CHOIX DE LA FONCTION DE PERTE :
    - Binaire  → BCEWithLogitsLoss
      Combine sigmoid + binary cross-entropy en une seule opération.
      Numériquement plus stable que d'appliquer sigmoid puis BCELoss.
      Permet de passer directement le logit brut (sortie de Linear).

    - Multi-classes → CrossEntropyLoss
      = LogSoftmax + NLLLoss. Calcule la perte pour N classes en une fois.
      Gère la normalisation softmax en interne → plus stable numériquement.

    Returns:
        dict avec "loss" et "accuracy" pour cette epoch
    """
    model.train()
    total_loss = 0.0
    correct    = 0
    total      = 0

    for images, labels in loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)

        if task == "binary":
            # outputs : [batch, 1] → squeeze → [batch]
            # labels  : [batch] entiers → float pour BCEWithLogitsLoss
            loss = criterion(outputs.squeeze(1), labels.float())
            preds = (torch.sigmoid(outputs.squeeze(1)) > 0.5).long()
        else: 
            loss = criterion(outputs, labels)
            preds = outputs.argmax(dim=1)
                        
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * images.size(0)
        correct    += (preds == labels).sum().item()
        total      += images.size(0)

    return {
        "loss": total_loss / total,
        "accuracy": correct / total,
    }


def evaluate(
    model: nn.Module,
    loader: torch.utils.data.DataLoader,
    criterion: nn.Module,
    device: torch.device,
    task: str = "binary",
    class_names: list[str] = None,
) -> dict:
    """
    Évalue le modèle sur un DataLoader (val ou test).
    Retourne toutes les métriques sans les afficher ni les sauvegarder.
    """
    model.eval()
    total_loss  = 0.0
    all_preds   = []
    all_labels  = []
    all_probs   = []  # pour la courbe ROC

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)

            if task == "binary":
                loss  = criterion(outputs.squeeze(1), labels.float())
                probs = torch.sigmoid(outputs.squeeze(1))
                preds = (probs > 0.5).long()
                all_probs.extend(probs.cpu().numpy())
            else: 
                loss = criterion(outputs, labels)
                preds = outputs.argmax(dim=1)
                probs = torch.softmax(outputs, dim=1)
                all_probs.extend(probs.cpu().numpy())

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

            total_loss += loss.item() * images.size(0)

    n = len(all_labels)
    return {
        "loss"        : total_loss / n,
        "accuracy"    : np.mean(np.array(all_preds) == np.array(all_labels)),
        "all_preds"   : np.array(all_preds),
        "all_labels"  : np.array(all_labels),
        "all_probs"   : np.array(all_probs),
        "class_names" : class_names or [],
    }


# ─────────────────────────────────────────────────────────────────
# BOUCLE PRINCIPALE D'ENTRAÎNEMENT (PHASES 1 + 2)
# ─────────────────────────────────────────────────────────────────

def train_full(
    model: nn.Module,
    train_loader,
    val_loader,
    device: torch.device,
    task: str = "binary",
    epochs_phase1: int = 10,
    epochs_phase2: int = 10,
    output_dir: str = "outputs",
    architecture: str = "mobilenet_v2",
) -> dict:
    """
    Entraîne le modèle en deux phases (feature extraction + fine-tuning).

    STRATÉGIE D'ARRÊT PRÉCOCE (Early Stopping) :
    Si la perte de validation ne s'améliore pas pendant 5 epochs consécutives,
    on arrête et on charge le meilleur modèle sauvegardé.
    POURQUOI : évite l'overfitting sur le petit dataset PHP.

    SCHEDULER DE LEARNING RATE :
    ReduceLROnPlateau : divise le lr par 2 si val_loss stagne 3 epochs.
    Complémentaire à l'early stopping : ralentit l'apprentissage avant d'arrêter.
    """
    # get_optimizer is now in the global scope
    # from module_a.model import get_optimizer

    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    if task == "binary":
        # Poids de classe pour gérer le déséquilibre : les faux négatifs
        # (accepter un fruit rejeté) coûtent plus cher à PHP que les faux positifs
        # pos_weight = nb_negatifs / nb_positifs → pénalise davantage les FN
        criterion = nn.BCEWithLogitsLoss(pos_weight=torch.tensor([2.0]).to(device))
        print("[Criterion] BCEWithLogitsLoss avec pos_weight=2.0")
        print("  → Un faux négatif (rejeté classé export) est 2x plus pénalisé")
        print("    qu'un faux positif (export classé rejeté). Adapté au risque PHP.")
    else:
        #criterion = nn.CrossEntropyLoss()
        #print("[Criterion] CrossEntropyLoss pour classification multi-classes")
        
        # Pour du multi-label, on utilise BCEWithLogitsLoss

        criterion = nn.CrossEntropyLoss()
        print("[Criterion] CrossEntropyLoss pour classification multi-classes")

    history = {"train_loss": [], "val_loss": [], "train_acc": [], "val_acc": []}
    best_val_loss    = float("inf")
    patience_counter = 0
    PATIENCE         = 5
    best_model_path  = output_dir / f"best_{architecture}_{task}.pth"

    # ── PHASE 1 : Feature Extraction ──────────────────────────────
    print(f"\n{'='*50}")
    print(f"PHASE 1 : Feature Extraction — {epochs_phase1} epochs")
    print(f"  Backbone gelé, seule la tête est entraînée")
    print(f"{'='*50}")

    optimizer = get_optimizer(model, phase=1)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode="min", factor=0.5, patience=3
    )

    for epoch in range(1, epochs_phase1 + 1):
        train_metrics = train_one_epoch(model, train_loader, optimizer, criterion, device, task)
        val_metrics   = evaluate(model, val_loader, criterion, device, task)

        scheduler.step(val_metrics["loss"])

        history["train_loss"].append(train_metrics["loss"])
        history["val_loss"].append(val_metrics["loss"])
        history["train_acc"].append(train_metrics["accuracy"])
        history["val_acc"].append(val_metrics["accuracy"])

        print(f"Epoch {epoch:02d}/{epochs_phase1} | "
              f"Train Loss: {train_metrics['loss']:.4f} Acc: {train_metrics['accuracy']:.3f} | "
              f"Val Loss: {val_metrics['loss']:.4f} Acc: {val_metrics['accuracy']:.3f}")

        if val_metrics["loss"] < best_val_loss:
            best_val_loss = val_metrics["loss"]
            torch.save(model.state_dict(), best_model_path)
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= PATIENCE:
                print(f"  Early stopping déclenché à l'epoch {epoch}")
                break

    # ── PHASE 2 : Fine-tuning ──────────────────────────────────────
    print(f"\n{'='*50}")
    print(f"PHASE 2 : Fine-tuning — {epochs_phase2} epochs")
    print(f"  Dernières couches backbone dégelées, lr réduit")
    print(f"{'='*50}")

    # Charger le meilleur modèle de la Phase 1
    model.load_state_dict(torch.load(best_model_path))
    # Dégeler les dernières couches
    if hasattr(model, "features"):
        # _unfreeze_last_n_layers is now in the global scope
        # from module_a.model import _unfreeze_last_n_layers
        n_unfreeze = 5 if "mobilenet" in architecture else 3
        _unfreeze_last_n_layers(model.features, n_unfreeze)

    optimizer = get_optimizer(model, phase=2)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode="min", factor=0.5, patience=3
    )
    patience_counter = 0

    for epoch in range(1, epochs_phase2 + 1):
        train_metrics = train_one_epoch(model, train_loader, optimizer, criterion, device, task)
        val_metrics   = evaluate(model, val_loader, criterion, device, task)

        scheduler.step(val_metrics["loss"])

        history["train_loss"].append(train_metrics["loss"])
        history["val_loss"].append(val_metrics["loss"])
        history["train_acc"].append(train_metrics["accuracy"])
        history["val_acc"].append(val_metrics["accuracy"])

        print(f"Epoch FT {epoch:02d}/{epochs_phase2} | "
              f"Train Loss: {train_metrics['loss']:.4f} Acc: {train_metrics['accuracy']:.3f} | "
              f"Val Loss: {val_metrics['loss']:.4f} Acc: {val_metrics['accuracy']:.3f}")

        if val_metrics["loss"] < best_val_loss:
            best_val_loss = val_metrics["loss"]
            torch.save(model.state_dict(), best_model_path)
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= PATIENCE:
                print(f"  Early stopping déclenché à l'epoch FT {epoch}")
                break

    # Charger le meilleur modèle final
    model.load_state_dict(torch.load(best_model_path))
    print(f"\nMeilleur modèle chargé depuis {best_model_path}")

    return {"model": model, "history": history, "best_val_loss": best_val_loss}


# ─────────────────────────────────────────────────────────────────
# MÉTRIQUES ET VISUALISATIONS (exigées par le cahier des charges)
# ─────────────────────────────────────────────────────────────────

def compute_inference_time(model: nn.Module, device: torch.device, n_runs: int = 100) -> float:
    """
    Mesure le temps d'inférence moyen par image en millisecondes.

    PERTINENCE POUR PHP : le tapis convoyeur traite 3-4 mains par minute.
    Une main = ~12-15 doigts. Soit ~1 image toutes les 1-2 secondes maximum.
    Le modèle doit inférer en < 200ms pour ne pas bloquer le convoyeur.
    MobileNetV2 (~30ms) et EfficientNet-B0 (~50ms) sont largement compatibles.
    """
    model.eval()
    dummy_input = torch.randn(1, 3, 224, 224).to(device)

    # Warm-up : les premières passes sont lentes (allocations mémoire GPU)
    with torch.no_grad():
        for _ in range(10):
            _ = model(dummy_input)

    # Mesure réelle
    times = []
    with torch.no_grad():
        for _ in range(n_runs):
            start = time.perf_counter()
            _ = model(dummy_input)
            if device.type == "cuda":
                torch.cuda.synchronize()  # attendre la fin GPU
            end = time.perf_counter()
            times.append((end - start) * 1000)  # en ms

    mean_ms = np.mean(times)
    std_ms  = np.std(times)
    print(f"\n[Inference Time] {mean_ms:.2f} ± {std_ms:.2f} ms/image")
    print(f"  → Compatible convoyeur PHP : {'OUI' if mean_ms < 200 else 'NON (> 200ms)'}")
    return mean_ms


def plot_confusion_matrix_php(
    all_labels: np.ndarray,
    all_preds: np.ndarray,
    class_names: list[str],
    task: str,
    save_path: str = None,
) -> None:
    """
    Génère et sauvegarde la matrice de confusion avec commentaire métier PHP.

    LECTURE DE LA MATRICE POUR LE JURY PHP :
    Pour la tâche BINAIRE (export=0, rejeté=1) :

    Matrice :         Prédit Export  Prédit Rejeté
    Réel Export   [[      TP,              FP      ]]
    Réel Rejeté   [[      FN,              TN      ]]

    ⚠ FN (Faux Négatif) = banane rejetée classée "export"
      → C'est la pire erreur pour PHP : risque de rejet de conteneur en Europe
      → Justifie pos_weight=2.0 dans BCEWithLogitsLoss

    ⚠ FP (Faux Positif) = bonne banane classée "rejetée"
      → Perte de revenu mais pas de risque client
      → Moins grave, mais à minimiser aussi pour la rentabilité
    """
    cm = confusion_matrix(all_labels, all_preds)

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Matrice brute (valeurs absolues)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)
    disp.plot(ax=axes[0], colorbar=False, cmap="Blues")
    axes[0].set_title("Matrice de confusion — valeurs absolues")

    # Matrice normalisée (pourcentages)
    cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)
    disp_norm = ConfusionMatrixDisplay(confusion_matrix=cm_norm, display_labels=class_names)
    disp_norm.plot(ax=axes[1], colorbar=False, cmap="Blues",
                   values_format=".1%")
    axes[1].set_title("Matrice de confusion — taux normalisés")

    # Annotation métier PHP pour la tâche binaire
    if task == "binary" and len(class_names) == 2:
        tn, fp, fn, tp = cm.ravel()
        php_comment = (
            f"Analyse métier PHP :\n"
            f"  ✓ Bananes export correctement acceptées (TP) : {tp}\n"
            f"  ✗ Bananes rejetées acceptées par erreur (FN) : {fn}  ← CRITIQUE\n"
            f"    → Risque rejet conteneur Europe\n"
            f"  ! Bonnes bananes rejetées par erreur (FP)     : {fp}\n"
            f"    → Perte de revenu estimée : {fp * 0.45:.0f} € (à 0.45€/banane)"
        )
        fig.text(0.5, -0.05, php_comment, ha="center", fontsize=9,
                 bbox=dict(boxstyle="round", facecolor="lightyellow", alpha=0.8))

    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches="tight")
        print(f"[Confusion Matrix] Sauvegardée : {save_path}")
    plt.close()


def plot_roc_curve(
    all_labels: np.ndarray,
    all_probs: np.ndarray,
    save_path: str = None,
) -> float:
    """
    Trace et sauvegarde la courbe ROC avec l'AUC.

    INTERPRÉTATION POUR LE JURY :
    AUC (Aire Sous la Courbe) mesure la capacité du modèle à discriminer
    les deux classes indépendamment du seuil choisi.

    AUC = 0.5 → modèle aléatoire (inutile)
    AUC = 0.7 → acceptable
    AUC = 0.9 → très bon
    AUC = 1.0 → parfait (suspect : probablement du data leakage)

    LIEN AVEC LE MDP (Module C) :
    La courbe ROC permet de choisir le seuil de décision optimal en
    fonction des coûts métier PHP (coût FN vs coût FP) → ce seuil
    sera intégré comme paramètre dans le MDP du Module C.
    """
    fpr, tpr, thresholds = roc_curve(all_labels, all_probs)
    roc_auc = auc(fpr, tpr)

    # Point de fonctionnement optimal : maximise TPR - FPR (index de Youden)
    optimal_idx = np.argmax(tpr - fpr)
    optimal_threshold = thresholds[optimal_idx]
    optimal_fpr = fpr[optimal_idx]
    optimal_tpr = tpr[optimal_idx]

    fig, ax = plt.subplots(figsize=(8, 6))
    ax.plot(fpr, tpr, color="steelblue", lw=2,
            label=f"Courbe ROC (AUC = {roc_auc:.3f})")
    ax.plot([0, 1], [0, 1], "k--", lw=1, label="Modèle aléatoire (AUC = 0.5)")
    ax.scatter(optimal_fpr, optimal_tpr, color="red", s=80, zorder=5,
               label=f"Seuil optimal = {optimal_threshold:.2f}")

    ax.set_xlabel("Taux de Faux Positifs (FPR)")
    ax.set_ylabel("Taux de Vrais Positifs (TPR = Rappel)")
    ax.set_title("Courbe ROC — Classification Binaire BananaVision\n"
                 f"(export vs rejeté) — AUC = {roc_auc:.3f}")
    ax.legend(loc="lower right")
    ax.grid(True, alpha=0.3)

    ax.annotate(
        f"Seuil optimal PHP = {optimal_threshold:.2f}\n"
        f"TPR = {optimal_tpr:.1%} | FPR = {optimal_fpr:.1%}",
        xy=(optimal_fpr, optimal_tpr), xytext=(0.4, 0.3),
        arrowprops=dict(arrowstyle="->", color="red"),
        fontsize=9, color="red",
    )

    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches="tight")
        print(f"[ROC Curve] Sauvegardée : {save_path}")
    plt.close()

    print(f"\n[AUC] {roc_auc:.4f}")
    print(f"[Seuil optimal] {optimal_threshold:.2f} "
          f"(TPR={optimal_tpr:.3f}, FPR={optimal_fpr:.3f})")
    return roc_auc


def print_classification_report_php(
    all_labels: np.ndarray,
    all_preds: np.ndarray,
    class_names: list[str],
    task: str,
) -> None:
    """
    Affiche le rapport de classification avec commentaires métier PHP.
    """
    print("\n" + "="*60)
    print(f"RAPPORT DE CLASSIFICATION — {task.upper()}")
    print("="*60)
    print(classification_report(all_labels, all_preds, target_names=class_names))

    if task == "binary":
        print("INTERPRÉTATION MÉTIER PHP :")
        print("  Précision 'rejeté' : parmi les fruits classés rejetés, combien le sont vraiment ?")
        print("    → Une faible précision = beaucoup de bonnes bananes rejetées (perte de revenu)")
        print("  Rappel 'rejeté'    : parmi les fruits vraiment rejetés, combien sont détectés ?")
        print("    → Un faible rappel = des fruits non-conformes acceptés (risque client Europe)")
        print("  F1-score = équilibre entre précision et rappel")
        print("  Pour PHP : prioriser le RAPPEL sur 'rejeté' (éviter les FN critiques)")

    #if task == "multiclass":
    # Convertir les prédictions et labels (one-hot -> indices pour le rapport)
    # Pour chaque classe, on calcule les métriques binaires
       # from sklearn.metrics import precision_recall_fscore_support
        #print("\n" + "="*60)
        #print("RAPPORT MULTI-LABEL (par défaut)")
        #print("="*60)

        #for i, name in enumerate(class_names):
            #y_true = all_labels[:, i] if all_labels.ndim > 1 else (all_labels == i).astype(int)
            #y_pred = all_preds[:, i]
            #p, r, f, _ = precision_recall_fscore_support(y_true, y_pred, average='binary')
            #print(f"{name:20}   précision={p:.3f}   rappel={r:.3f}   f1={f:.3f}")
```

```python
import argparse
import time
from pathlib import Path
import sys # Import sys
import os # Import os

# Add the current directory to sys.path to allow importing local modules
if os.getcwd() not in sys.path:
    sys.path.insert(0, os.getcwd())

import torch

# Functions are now directly in the global scope from executed cells
# from module_a.preprocessing import load_binary_dataset, load_multiclass_dataset
# from module_a.model import build_model
# from module_a.training import (
#     train_full,
#     evaluate,
#     compute_inference_time,
#     plot_confusion_matrix_php,
#     plot_roc_curve,
#     print_classification_report_php,
# )

def parse_args():
    parser = argparse.ArgumentParser(description="BananaVision — Module A CNN")
    parser.add_argument("--task",     choices=["binary", "multiclass"], required=True)
    parser.add_argument("--arch",     choices=["mobilenet_v2", "efficientnet_b0"],
                        default="efficientnet_b0")
    parser.add_argument("--data_dir", type=str, required=True,
                        help="Chemin vers le dossier du dataset")
    parser.add_argument("--batch_size",    type=int, default=32)
    parser.add_argument("--epochs_phase1", type=int, default=10)
    parser.add_argument("--epochs_phase2", type=int, default=10)
    parser.add_argument("--output_dir",    type=str, default="outputs/module_a")
    parser.add_argument("--compare",       action="store_true",
                        help="Compare MobileNetV2 vs EfficientNet-B0")
    return parser.parse_args()


def run_single_experiment(args, architecture: str) -> dict:
    """
    Exécute un entraînement complet pour une architecture donnée.
    Retourne les métriques finales pour comparaison.
    """
    print(f"\n{'#'*60}")
    print(f"# EXPÉRIENCE : {architecture.upper()} — Tâche : {args.task.upper()}")
    print(f"{'#'*60}")

    # ── 1. Détection du device ────────────────────────────────────
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"\n[Device] Utilisation de : {device}")
    if device.type == "cuda":
        print(f"  GPU : {torch.cuda.get_device_name(0)}")
        print(f"  VRAM disponible : {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")

    # ── 2. Chargement des données ─────────────────────────────────
    print(f"\n[Data] Chargement depuis {args.data_dir}")
    if args.task == "binary":
    # IMPORTANT : args.data_dir doit maintenant pointer vers le dossier parent
    # (par exemple './dataset/binary' et non './dataset/binary/train')
        train_loader, val_loader, test_loader, class_names = load_binary_dataset(
        data_dir=args.data_dir,
        batch_size=args.batch_size,
    )
    else:
        train_loader, val_loader, test_loader, class_names = load_multiclass_dataset(
            data_dir=args.data_dir,
            batch_size=args.batch_size,
        )
    num_classes = 1 if args.task == "binary" else len(class_names)

    # ── 3. Construction du modèle (Phase 1 : backbone gelé) ───────
    print(f"\n[Model] Chargement {architecture} pré-entraîné (ImageNet)")
    model = build_model(
        architecture=architecture,
        num_classes=num_classes,
        task=args.task,
        freeze_backbone=True,   # commence avec backbone gelé
    ).to(device)

    # ── 4. Entraînement en 2 phases ───────────────────────────────
    print(f"\n[Training] Démarrage de l'entraînement en 2 phases")
    result = train_full(
        model=model,
        train_loader=train_loader,
        val_loader=val_loader,
        device=device,
        task=args.task,
        epochs_phase1=args.epochs_phase1,
        epochs_phase2=args.epochs_phase2,
        output_dir=args.output_dir,
        architecture=architecture,
    )
    model = result["model"]

    # ── 5. Évaluation finale sur le TEST SET ──────────────────────
    # IMPORTANT : le test set ne doit JAMAIS avoir été vu pendant l'entraînement
    # C'est la seule mesure honnête des performances réelles sur de nouvelles bananes
    print(f"\n[Evaluation] Test set — données jamais vues pendant l'entraînement")

    import torch.nn as nn
    if args.task == "binary":
        criterion = nn.BCEWithLogitsLoss(pos_weight=torch.tensor([2.0]).to(device))
    else:
        criterion = nn.CrossEntropyLoss()

    test_metrics = evaluate(
        model, test_loader, criterion, device, args.task, class_names
    )

    # ── 6. Rapport de classification ─────────────────────────────
    print_classification_report_php(
        test_metrics["all_labels"],
        test_metrics["all_preds"],
        class_names,
        args.task,
    )

    # ── 7. Matrice de confusion ───────────────────────────────────
    output_dir = Path(args.output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    cm_path = output_dir / f"confusion_matrix_{architecture}_{args.task}.png"
    plot_confusion_matrix_php(
        test_metrics["all_labels"],
        test_metrics["all_preds"],
        class_names,
        args.task,
        save_path=str(cm_path),
    )

    # ── 8. Courbe ROC (binaire uniquement) ───────────────────────
    roc_auc = None
    if args.task == "binary":
        roc_path = output_dir / f"roc_curve_{architecture}.png"
        roc_auc = plot_roc_curve(
            test_metrics["all_labels"],
            test_metrics["all_probs"],
            save_path=str(roc_path),
        )

    # ── 9. Temps d'inférence ──────────────────────────────────────
    inference_ms = compute_inference_time(model, device)

    # ── 10. Résumé ───────────────────────────────────────────────
    summary = {
        "architecture" : architecture,
        "task"         : args.task,
        "test_accuracy": test_metrics["accuracy"],
        "auc"          : roc_auc,
        "inference_ms" : inference_ms,
    }

    print(f"\n{'='*50}")
    print(f"RÉSUMÉ FINAL — {architecture}")
    print(f"{'='*50}")
    print(f"  Accuracy (test)  : {test_metrics['accuracy']:.3f}")
    if roc_auc:
        print(f"  AUC (test)       : {roc_auc:.3f}")
    print(f"  Inference time   : {inference_ms:.1f} ms/image")
    print(f"  Figures sauvées  : {output_dir}/")

    return summary


def compare_architectures(args) -> None:
    """
    Compare MobileNetV2 et EfficientNet-B0 sur les mêmes données.
    Produit le tableau comparatif pour le rapport technique (Section A2).
    """
    results = {}
    for arch in ["mobilenet_v2", "efficientnet_b0"]:
        results[arch] = run_single_experiment(args, arch)

    print(f"\n{'='*60}")
    print("TABLEAU COMPARATIF — MobileNetV2 vs EfficientNet-B0")
    print(f"{'='*60}")
    print(f"{'Métrique':<25} {'MobileNetV2':>15} {'EfficientNet-B0':>18}")
    print("-"*60)

    metrics_labels = {
        "test_accuracy": "Accuracy (test)",
        "auc"          : "AUC ROC",
        "inference_ms" : "Temps inférence (ms)",
    }
    for key, label in metrics_labels.items():
        v1 = results["mobilenet_v2"][key]
        v2 = results["efficientnet_b0"][key]
        if v1 is not None and v2 is not None:
            better = "MobileNetV2" if v1 < v2 else "EfficientNet-B0"
            if key == "inference_ms":
                better = "MobileNetV2" if v1 < v2 else "EfficientNet-B0"
                print(f"{label:<25} {v1:>14.1f}ms {v2:>17.1f}ms  ← {better} plus rapide")
            else:
                print(f"{label:<25} {v1:>15.4f} {v2:>18.4f}  ← {better} meilleur")

    print(f"\nRECOMMANDATION POUR LE RAPPORT :")
    m1 = results["mobilenet_v2"]
    m2 = results["efficientnet_b0"]
    if m1["test_accuracy"] is not None and m2["test_accuracy"] is not None:
        diff_acc = abs(m1["test_accuracy"] - m2["test_accuracy"])
        if diff_acc < 0.03:
            print(f"  Les deux architectures ont des performances similaires (Δ={diff_acc:.1%}).")
            print(f"  Choisir MobileNetV2 pour le déploiement temps réel sur convoyeur PHP.")
        elif m2["test_accuracy"] > m1["test_accuracy"]:
            print(f"  EfficientNet-B0 est significativement meilleur (+{diff_acc:.1%}).")
            print(f"  Justifier le coût en inférence vs gain de précision dans le rapport.")


if __name__ == "__main__":
    # Save original sys.argv
    original_argv = sys.argv
    try:
        # Simulate command-line arguments for Colab execution
        # Replace with your desired arguments, e.g., for binary classification
        sys.argv = ['run_module_a.py', '--task', 'binary', '--data_dir', './dataset/binary']
        # Or for multiclass:
        # sys.argv = ['run_module_a.py', '--task', 'multiclass', '--data_dir', './dataset/multiclass', '--arch', 'efficientnet_b0']
        # Or for comparison:
        # sys.argv = ['run_module_a.py', '--task', 'binary', '--data_dir', './dataset/binary', '--compare']

        args = parse_args()

        if args.compare:
            compare_architectures(args)
        else:
            run_single_experiment(args, args.arch)
    finally:
        # Restore original sys.argv
        sys.argv = original_argv
```

```python
# ============================================================================
# EXTRACTION DES EMBEDDINGS POUR LE CLUSTERING (Module B)
# ============================================================================
import torch
import torch.nn as nn
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader
from pathlib import Path
import numpy as np
from tqdm import tqdm

# 1. Configuration
# On choisit le modèle binaire (ou multi‑classes) pour extraire les features
# Ici on prend le binaire car il a été entraîné sur plus de données (export/rejeté)
model_weights = "./outputs/module_a/best_efficientnet_b0_binary.pth"
# Dossier contenant les images pour le clustering (par exemple tout le dataset multiclass train/val/test)
# On va utiliser le dossier "dataset/multiclass/train" qui contient les trois classes de défauts
# (on peut aussi utiliser un mélange de export + défauts)
data_dir = Path("./dataset/multiclass/train")  # à adapter si besoin
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 2. Transformations (sans augmentation, juste resize + normalisation)
transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# 3. Chargement du dataset
dataset = datasets.ImageFolder(root=data_dir, transform=transform)
loader = DataLoader(dataset, batch_size=64, shuffle=False, num_workers=4)

# 4. Chargement du modèle et modification pour extraire les embeddings
def get_embedding_model():
    model = models.efficientnet_b0(weights=None)
    # Remplacer la tête de classification par une sortie d'embedding (avant le classifier)
    # Pour EfficientNet, les features sont dans model.classifier[1].in_features
    # On va créer un modèle qui retourne la sortie avant la dernière couche linéaire
    num_ftrs = model.classifier[1].in_features
    # On garde tout sauf le dernier Linear
    # On va plutôt utiliser la sortie de model.features + une moyenne globale
    # Mais plus simple : on enlève le classifier et on ajoute un AdaptiveAvgPool2d(1)
    # EfficientNet a déjà une moyenne globale dans son forward (après features)
    # On peut simplement faire forward, puis prendre la sortie avant classifier
    # Je préfère créer un hook ou modifier l'architecture
    # Solution : on remplace classifier par Identity et on récupère les features
    class FeatureExtractor(nn.Module):
        def __init__(self, original_model):
            super().__init__()
            self.features = original_model.features
            self.avgpool = original_model.avgpool
            # Pas de classifier
        def forward(self, x):
            x = self.features(x)
            x = self.avgpool(x)
            x = torch.flatten(x, 1)
            return x
    feature_model = FeatureExtractor(model)
    # Charger les poids du modèle entraîné (attention : le classifier a une couche supplémentaire)
    # On charge les poids du modèle complet, mais on ignore le classifier
    state_dict = torch.load(model_weights, map_location=device)
    # Supprimer les clés du classifier (car notre feature model n'a pas de classifier)
    new_state_dict = {k: v for k, v in state_dict.items() if not k.startswith('classifier')}
    feature_model.load_state_dict(new_state_dict, strict=False)
    feature_model.to(device)
    feature_model.eval()
    return feature_model

feature_model = get_embedding_model()

# 5. Extraction des embeddings
embeddings = []
labels = []
filenames = []

with torch.no_grad():
    for images, targets in tqdm(loader, desc="Extraction des embeddings"):
        images = images.to(device)
        emb = feature_model(images)  # shape [batch, 1280]
        embeddings.append(emb.cpu().numpy())
        labels.extend(targets.numpy())
        # Récupérer les chemins des images (optionnel)
        for i in range(len(images)):
            filenames.append(dataset.samples[len(embeddings)-1][0])  # approximatif

embeddings = np.vstack(embeddings)
labels = np.array(labels)
print(f"Embeddings extraits : shape {embeddings.shape}")
print(f"Nombre de classes : {len(dataset.classes)}")
print(f"Classes : {dataset.classes}")

# 6. Sauvegarde des embeddings pour une utilisation ultérieure
np.save("embeddings_multi.npy", embeddings)
np.save("labels_multi.npy", labels)
import json
with open("class_names_multi.json", "w") as f:
    json.dump(dataset.classes, f)
print("Embeddings sauvegardés dans embeddings_multi.npy")

# 7. Option : visualisation rapide avec PCA pour se donner une idée
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)
embeddings_2d = pca.fit_transform(embeddings)

plt.figure(figsize=(10, 8))
for i, class_name in enumerate(dataset.classes):
    idx = labels == i
    plt.scatter(embeddings_2d[idx, 0], embeddings_2d[idx, 1], label=class_name, alpha=0.5)
plt.legend()
plt.title("PCA des embeddings - Module B")
plt.xlabel("Composante 1")
plt.ylabel("Composante 2")
plt.savefig("pca_embeddings.png")
plt.show()
```

```python
# ============================================================================
# MODULE B : CLUSTERING NON SUPERVISÉ (K-MEANS) DES EMBEDDINGS CNN
# ============================================================================
# Objectif : Regrouper les images de bananes en familles de défauts naturels,
# sans utiliser les étiquettes. On utilise les caractéristiques (embeddings)
# extraites de la couche avant-dernière d'un CNN pré-entraîné (EfficientNet-B0).
# Le clustering permet de découvrir des sous-groupes que le CNN n'a pas appris
# explicitement, et d'identifier des catégories de défauts cohérentes avec le
# métier PHP.
# ============================================================================

import torch
import torch.nn as nn
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader
from pathlib import Path
import numpy as np
from tqdm import tqdm
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import json

# ----------------------------------------------------------------------------
# 1. CONFIGURATION
# ----------------------------------------------------------------------------
# On utilise le modèle binaire (export/rejeté) car il a été entraîné sur un grand
# nombre d'images et ses embeddings sont discriminants. On pourrait aussi utiliser
# le modèle multi-classes.
model_weights = "./outputs/module_a/best_efficientnet_b0_binary.pth"
# Dossier contenant les images à clusteriser (par exemple toutes les images du
# dataset multi-classes : défauts mécaniques, fongiques, mûres). On peut aussi
# ajouter des images "export" pour voir comment elles se regroupent.
data_dir = Path("./dataset/multiclass/train")  # à adapter selon votre structure
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ----------------------------------------------------------------------------
# 2. TRANSFORMATIONS (standard pour EfficientNet)
# ----------------------------------------------------------------------------
# Les images sont redimensionnées et normalisées comme pour l'entraînement.
# Pas d'augmentation car on veut des embeddings stables.
transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# ----------------------------------------------------------------------------
# 3. CHARGEMENT DES IMAGES
# ----------------------------------------------------------------------------
dataset = datasets.ImageFolder(root=data_dir, transform=transform)
loader = DataLoader(dataset, batch_size=64, shuffle=False, num_workers=4)
print(f"Nombre total d'images : {len(dataset)}")
print(f"Classes (pour information) : {dataset.classes}")

# ----------------------------------------------------------------------------
# 4. MODÈLE POUR EXTRAIRE LES EMBEDDINGS
# ----------------------------------------------------------------------------
# Principe : On supprime la tête de classification (dernière couche linéaire)
# du CNN, et on récupère la sortie de la couche précédente (les features).
# Pour EfficientNet-B0, la dimension des embeddings est de 1280.
# Ces embeddings représentent une description dense et abstraite de l'image,
# apprise par le réseau sur ImageNet puis affinée sur nos bananes.
class FeatureExtractor(nn.Module):
    """Extracteur d'embeddings : même backbone que le modèle entraîné,
    mais sans le dernier classifieur."""
    def __init__(self, original_model):
        super().__init__()
        self.features = original_model.features
        self.avgpool = original_model.avgpool   # global average pooling
    def forward(self, x):
        x = self.features(x)
        x = self.avgpool(x)
        x = torch.flatten(x, 1)   # vecteur 1D de 1280 dimensions
        return x

# Charger le modèle EfficientNet-B0 sans ses poids (on va charger les poids
# du modèle entraîné, mais en ignorant la dernière couche).
model = models.efficientnet_b0(weights=None)
feature_model = FeatureExtractor(model)

# Charger les poids entraînés (seulement ceux qui existent dans le backbone)
state_dict = torch.load(model_weights, map_location=device)
# On supprime les clés du classifieur (car notre extracteur n'a pas de classifieur)
filtered_state = {k: v for k, v in state_dict.items() if not k.startswith('classifier')}
feature_model.load_state_dict(filtered_state, strict=False)
feature_model.to(device)
feature_model.eval()
print("Modèle extracteur prêt.")

# ----------------------------------------------------------------------------
# 5. EXTRACTION DES EMBEDDINGS
# ----------------------------------------------------------------------------
embeddings = []
labels = []   # étiquettes d'origine (classe réelle : defaut_mecanique, etc.)

with torch.no_grad():
    for images, targets in tqdm(loader, desc="Extraction des embeddings"):
        images = images.to(device)
        emb = feature_model(images)        # shape: [batch, 1280]
        embeddings.append(emb.cpu().numpy())
        labels.extend(targets.numpy())

embeddings = np.vstack(embeddings)          # [N, 1280]
labels = np.array(labels)                  # [N]
print(f"Embeddings extraits : shape {embeddings.shape}")

# Sauvegarde pour réutilisation ultérieure
np.save("embeddings.npy", embeddings)
np.save("labels.npy", labels)
with open("class_names.json", "w") as f:
    json.dump(dataset.classes, f)

# ----------------------------------------------------------------------------
# 6. VISUALISATION PAR PCA
# ----------------------------------------------------------------------------
# PCA (Analyse en Composantes Principales) : technique de réduction de dimension
# linéaire qui projette les données sur les axes de variance maximale.
# On l'utilise pour visualiser les embeddings en 2D.
pca = PCA(n_components=2)
embeddings_2d = pca.fit_transform(embeddings)

plt.figure(figsize=(10, 8))
for i, class_name in enumerate(dataset.classes):
    idx = labels == i
    plt.scatter(embeddings_2d[idx, 0], embeddings_2d[idx, 1],
                label=class_name, alpha=0.6, s=10)
plt.legend()
plt.title("Projection PCA des embeddings (Module B)")
plt.xlabel("Première composante principale")
plt.ylabel("Deuxième composante principale")
plt.savefig("pca_embeddings.png", dpi=150)
plt.show()

# ----------------------------------------------------------------------------
# 7. CLUSTERING K-MEANS : DÉTERMINATION DU NOMBRE DE CLUSTERS k
# ----------------------------------------------------------------------------
# K-means : algorithme de partitionnement qui regroupe les points en k clusters
# en minimisant la variance intra-cluster. Il faut choisir k.
# Deux méthodes courantes : le "coude" sur l'inertie et le score silhouette.
# - Inertie : somme des distances au carré des points à leur centroïde.
#   Diminue avec k ; on cherche un coude (brusque changement de pente).
# - Score silhouette : mesure de cohésion intra-cluster et séparation inter-cluster.
#   Entre -1 et 1, plus élevé est mieux.
K_range = range(2, 11)
inertias = []
silhouette_scores = []

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(embeddings)
    inertias.append(kmeans.inertia_)
    score = silhouette_score(embeddings, kmeans.labels_)
    silhouette_scores.append(score)
    print(f"k={k}: inertie={kmeans.inertia_:.2f}, silhouette={score:.3f}")

# Tracé des courbes
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.plot(K_range, inertias, 'bo-')
ax1.set_xlabel('Nombre de clusters k')
ax1.set_ylabel('Inertie')
ax1.set_title('Méthode du coude')
ax1.grid(True)

ax2.plot(K_range, silhouette_scores, 'ro-')
ax2.set_xlabel('Nombre de clusters k')
ax2.set_ylabel('Score silhouette')
ax2.set_title('Score silhouette (meilleur = k optimal)')
ax2.grid(True)
plt.tight_layout()
plt.savefig("kmeans_selection.png")
plt.show()

# Choisir le k qui maximise le score silhouette
best_k = K_range[np.argmax(silhouette_scores)]
print(f"✅ Nombre de clusters optimal selon silhouette : {best_k}")

# ----------------------------------------------------------------------------
# 8. APPLICATION DE K-MEANS AVEC LE MEILLEUR k
# ----------------------------------------------------------------------------
kmeans_final = KMeans(n_clusters=best_k, random_state=42, n_init=10)
clusters = kmeans_final.fit_predict(embeddings)

# ----------------------------------------------------------------------------
# 9. ANALYSE DE LA COMPOSITION DES CLUSTERS
# ----------------------------------------------------------------------------
# On regarde, pour chaque cluster, la répartition des classes réelles (annotations).
# Cela permet d'interpréter ce que représente chaque cluster en termes de défaut.
print("\n--- Composition des clusters ---")
for cluster_id in range(best_k):
    idx = clusters == cluster_id
    nb = np.sum(idx)
    print(f"\nCluster {cluster_id} : {nb} images")
    # Distribution des classes réelles
    unique, counts = np.unique(labels[idx], return_counts=True)
    for cls, cnt in zip(unique, counts):
        print(f"   - {dataset.classes[cls]}: {cnt} ({100*cnt/nb:.1f}%)")

# On peut aussi sauvegarder les clusters pour le Module C
np.save("clusters.npy", clusters)
# Créer un mapping cluster -> interprétation (à définir selon les résultats)
# Par exemple, si le cluster 0 est majoritairement "defaut_mecanique", on peut
# lui associer une action recommandée (transformation, etc.)
print("\n✅ Clustering terminé. Les clusters sont sauvegardés dans 'clusters.npy'.")

# ----------------------------------------------------------------------------
# NOTE POUR LE MODULE C (MDP)
# ----------------------------------------------------------------------------
# Les clusters obtenus peuvent remplacer les classes de défaut comme "états"
# dans le MDP. Il faudra alors estimer les probabilités de transition
# entre clusters (ou entre cluster et action). Ceci sera fait ultérieurement.
```

```python
# =============================================================================
# MODULE C : PROCESSUS DE DÉCISION DE MARKOV (MDP)
# =============================================================================
# Objectif : Définir une politique de tri optimale pour PHP en tenant compte
# des erreurs de classification, des gains/coûts et de l'évolution future.
# États : 4 (Export, defaut_mecanique, maladie_fongique, mure)
# Actions : 4 (Accepter Export, Accepter Catégorie II, Rediriger Transformation, Contrôle manuel)
# =============================================================================

import numpy as np

# -----------------------------------------------------------------------------
# 1. DÉFINITION DES ÉTATS, ACTIONS, TRANSITIONS ET RÉCOMPENSES
# -----------------------------------------------------------------------------
state_names = ["Export", "defaut_mecanique", "maladie_fongique", "mure"]
n_states = len(state_names)

action_names = [
    "Accepter Export",
    "Accepter Catégorie II",
    "Rediriger Transformation",
    "Suspendre contrôle manuel"
]
n_actions = len(action_names)

# Matrice de transition P(s' | s) – issue de la confusion du CNN multi-classes
# Lignes = état réel, colonnes = état observé (nouvel état)
transition = np.array([
    [0.99, 0.003, 0.003, 0.004],  # Export
    [0.00, 1.00, 0.00, 0.00],     # defaut_mecanique
    [0.00, 0.00, 0.53, 0.47],     # maladie_fongique
    [0.00, 0.00, 0.00, 1.00]      # mure
])

# Récompenses R(s, a) en euros par main (estimations – à adapter)
R = np.array([
    [ 1.96,  1.40,  0.40, -0.10],  # Export
    [-2.50,  1.40,  0.40, -0.10],  # defaut_mecanique
    [-2.50,  1.40,  0.20, -0.10],  # maladie_fongique
    [-2.50,  1.40,  0.40, -0.10]   # mure
])

gamma = 0.95   # facteur d'actualisation

# -----------------------------------------------------------------------------
# 2. ALGORITHMES DE RÉSOLUTION
# -----------------------------------------------------------------------------
def value_iteration(trans, rewards, gamma, theta=1e-6):
    """Value Iteration : itération sur les valeurs des états."""
    V = np.zeros(n_states)
    while True:
        delta = 0
        for s in range(n_states):
            v = V[s]
            # Calcul de la Q‑valeur pour chaque action
            Q = [rewards[s, a] + gamma * np.sum(trans[s, :] * V) for a in range(n_actions)]
            V[s] = max(Q)
            delta = max(delta, abs(v - V[s]))
        if delta < theta:
            break
    # Politique déduite
    policy = np.zeros(n_states, dtype=int)
    for s in range(n_states):
        Q = [rewards[s, a] + gamma * np.sum(trans[s, :] * V) for a in range(n_actions)]
        policy[s] = np.argmax(Q)
    return V, policy

def policy_iteration(trans, rewards, gamma):
    """Policy Iteration : amélioration et évaluation alternées de la politique."""
    policy = np.zeros(n_states, dtype=int)   # politique initiale (action 0)
    while True:
        # Évaluation : résolution de (I - γ P) V = R
        A = np.eye(n_states) - gamma * trans
        b = np.array([rewards[s, policy[s]] for s in range(n_states)])
        V = np.linalg.solve(A, b)
        # Amélioration
        stable = True
        for s in range(n_states):
            old = policy[s]
            Q = [rewards[s, a] + gamma * np.sum(trans[s, :] * V) for a in range(n_actions)]
            policy[s] = np.argmax(Q)
            if old != policy[s]:
                stable = False
        if stable:
            break
    return V, policy

# -----------------------------------------------------------------------------
# 3. EXÉCUTION ET AFFICHAGE DES RÉSULTATS
# -----------------------------------------------------------------------------
V_vi, policy_vi = value_iteration(transition, R, gamma)
V_pi, policy_pi = policy_iteration(transition, R, gamma)

print("\n" + "="*60)
print("POLITIQUE OPTIMALE - VALUE ITERATION")
print("="*60)
for s, name in enumerate(state_names):
    print(f"État '{name:20}' -> {action_names[policy_vi[s]]}")

print("\n" + "="*60)
print("POLITIQUE OPTIMALE - POLICY ITERATION")
print("="*60)
for s, name in enumerate(state_names):
    print(f"État '{name:20}' -> {action_names[policy_pi[s]]}")

print("\n--- Comparaison des valeurs d'état ---")
print(f"{'État':<20} {'Value Iteration':>18} {'Policy Iteration':>18}")
for s, name in enumerate(state_names):
    print(f"{name:<20} {V_vi[s]:16.4f} {V_pi[s]:16.4f}")

print("\n--- Sensibilité à γ (Value Iteration) ---")
for g in [0.9, 0.95, 0.99]:
    V, _ = value_iteration(transition, R, g)
    print(f"γ = {g} : V = {V.round(3)}")
```

# Explications du Module C – Processus de Décision de Markov

## Objectif
Le MDP aide PHP Plantations à choisir automatiquement la meilleure action pour chaque banane en tenant compte :
- des **incertitudes** du modèle de classification (erreurs de confusion)
- des **gains et coûts** économiques (prix de vente, transformation, contrôle manuel)
- de l’**évolution future** (les décisions d’aujourd’hui influencent l’état des lots suivants).

## Éléments du MDP

### États (4)
Ce sont les « vérités terrain » (ce que le fruit est réellement) :
- `Export` : banane conforme, acceptable pour l’export.
- `defaut_mecanique` : fruit présentant un défaut d’origine mécanique.
- `maladie_fongique` : fruit attaqué par une maladie (cercosporiose, etc.).
- `mure` : fruit trop mûr.

### Actions (4)
- `Accepter Export` : envoyer le fruit à l’export (prix élevé, mais pénalité si le fruit est en réalité rejeté).
- `Accepter Catégorie II` : commercialisation sur le marché local (gain modéré).
- `Rediriger Transformation` : orienter vers la farine ou le jus (gain faible).
- `Suspendre contrôle manuel` : retirer la banane de la chaîne pour un contrôle humain (coût fixe, mais évite les erreurs graves).

### Matrice de transition \(P(s’ | s)\)
Elle indique la probabilité que le fruit, ayant un **état réel** \(s\), soit classé par le CNN dans un **état observé** \(s’\) (qui devient le nouvel état pour l’étape suivante).  
Elle est déduite de la **matrice de confusion du modèle multi‑classes** (normalisée).  
*Exemple* : une banane réellement *maladie fongique* a 53 % de chances d’être correctement identifiée (reste fongique) et 47 % d’être confondue avec *mure*.

### Récompenses \(R(s, a)\)
Gain (ou coût) immédiat en euros par main pour chaque couple (état, action).  
Les valeurs sont estimées à partir de données économiques publiques (FAO, BEAC) et d’hypothèses internes. **Elles doivent être remplacées par les chiffres réels de PHP**.

## Algorithmes de résolution

**Value Iteration**  
- On calcule itérativement la **valeur** \(V(s)\) de chaque état, c’est‑à‑dire le gain total actualisé qu’on peut espérer en partant de cet état.
- À chaque étape, on met à jour \(V(s)\) par l’équation de Bellman :
  \[
  V(s) = \max_a \left[ R(s,a) + \gamma \sum_{s'} P(s'|s) V(s') \right]
  \]
- On s’arrête quand les valeurs ne changent plus (convergence).
- La politique est ensuite déduite en choisissant l’action qui maximise cette expression.

**Policy Iteration**  
- On part d’une politique initiale (par exemple toujours l’action 0).
- On **évalue** cette politique en résolvant un système linéaire pour obtenir \(V\).
- On **améliore** la politique en choisissant l’action qui maximise la Q‑valeur pour chaque état.
- On répète jusqu’à ce que la politique soit stable.

Les deux méthodes donnent la **politique optimale**.

## Résultats

| État                  | Action choisie (Value Iteration) |
|-----------------------|----------------------------------|
| Export                | Accepter Export                  |
| defaut_mecanique      | Accepter Catégorie II            |
| maladie_fongique      | Accepter Catégorie II            |
| mure                  | Accepter Catégorie II            |

**Interprétation**  
- Les fruits “Export” sont envoyés à l’export (gain maximal).  
- Tous les fruits ayant un défaut sont orientés vers la **catégorie II** (consommation locale) plutôt que la transformation.  
  - La transformation rapporte moins (0,40 € contre 1,40 €) et le coût du contrôle manuel (–0,10 €) n’est pas justifié par le risque d’erreur sur la maladie fongique.
- Le contrôle manuel n’est jamais choisi dans cette configuration car trop coûteux ; il le deviendrait si la pénalité d’export d’un fruit rejeté augmentait.

**Analyse de sensibilité**  
En faisant varier \(\gamma\) (facteur d’actualisation), la politique reste inchangée ; seules les valeurs absolues changent légèrement. Le modèle est robuste.

## Limites et recommandations
- Les récompenses doivent être **actualisées** avec les prix et coûts réels de PHP Plantations.
- La transition pour la maladie fongique (0,53 / 0,47) devrait être améliorée en collectant plus d’images de cette classe.
- Le contrôle manuel pourrait être intégré comme action de secours si l’incertitude dépasse un seuil (POMDP).
